```python
from pathlib import Path
import os
import subprocess

REPO = Path('/home/azureuser/cloudfiles/code/Users/mooredouglas45/water-intel-platform')
os.chdir(REPO)
print(f'Working directory: {Path.cwd()}')

def run_cmd(cmd: str) -> int:
    print(f'\n$ {cmd}')
    p = subprocess.run(cmd, shell=True, text=True, capture_output=True)
    if p.stdout:
        print(p.stdout)
    if p.stderr:
        print(p.stderr)
    print(f'Exit code: {p.returncode}')
    return p.returncode
```
